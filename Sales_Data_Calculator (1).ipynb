# Sales Data Calculator

This notebook analyzes a manually created collection of sales values and calculates total sales, average sales, highest sale, and lowest sale.

**Tools:** Python, Jupyter Notebook

```python
# Create a list of sales values
sales = [12000, 15000, 11000, 18000, 22000, 17000, 14000, 25000, 19000, 16000]
sales
```

```python
# Calculate sales statistics
total_sales = sum(sales)
average_sales = total_sales / len(sales)
highest_sale = max(sales)
lowest_sale = min(sales)

print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("Highest Sale:", highest_sale)
print("Lowest Sale:", lowest_sale)
```

```python
# Display the results in a clear format
print(f"Total Sales   : ₹{total_sales:,.2f}")
print(f"Average Sales : ₹{average_sales:,.2f}")
print(f"Highest Sale  : ₹{highest_sale:,.2f}")
print(f"Lowest Sale   : ₹{lowest_sale:,.2f}")
```

## Result Explanation

- **Total sales:** ₹169,000.00 — the combined value of all 10 sales.
- **Average sales:** ₹16,900.00 — the mean value of a sale.
- **Highest sale:** ₹25,000.00 — the largest value in the list.
- **Lowest sale:** ₹11,000.00 — the smallest value in the list.

Python's built-in `sum()`, `max()`, and `min()` functions make these calculations simple and readable.
